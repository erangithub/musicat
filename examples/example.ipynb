```python
C = major('c',1)

# ---------------------------------------------------------------- right hand

# bars 1-4: two two-bar phrases
theme = (deg ('1 3 5 0 1 2 1') @ rhy(    '2 4 4  4.  16 16 4 r4') \
         + deg ('6 5 8 5 4 3 4 3') @ rhy('2 4 4  4 8 16 16 4 r4')).trans('8')

# bars 5-8: one scale figure, a step lower each bar
run1 = deg('1 0 -1 -2') @ (deg('6..13..6') @ rhy('8 [ 16'))
run2 = deg('2..8#..9 6..8#..12') @ rhy('8 [ 16')
run3 = deg('1 -1') @ deg('13..15..10') @ rhy('16') 
run4 = deg('7 12,,8 9 12,,8') @ rhy('8')

runs = run1 + run2 + run3 + run4

right_hand = (theme + runs) >> C

# ----------------------------------------------------------------- left hand

# bars 1-4: an Alberti bass. The figure is low-high-middle-high, so it is the
# same degrees over each chord; the chord is the palette.
alberti = deg('1 3 2 3') @ rhy('8')
tonic = '1 3 5'
chords = [tonic, tonic, '2 4 5', tonic, '1 4 6', tonic, '0 2 5', tonic]
alberti_bars = sum((alberti >> deg(chord) for chord in chords), Idea())

# bars 5-8: bass note and a dyad on the fourth beat
steps = (deg('4 <-4 1>   <-5 1> <-5 1>   <-6 1> <-6 0>   <-7 1> <-7 -5>')
         @ rhy('4 r4 r4 4'))

left_hand = (alberti_bars + steps) >> C


def score():
    return Score([Part(right_hand, 'Right hand'), Part(left_hand, 'Left hand')], '4/4')

score()

```

```python
from solfa import deg, rhy

f = deg('1 2 3 4')
rhy('r4') + f[2:4] + rhy('r8') 
g = deg('8 7 8')

def insert(idea1, idea2, index):
    return idea1[:index] + idea2 + idea1[index:]

insert(f, g, 2) @ rhy('8') @ rhy('4 [ 8')
```
